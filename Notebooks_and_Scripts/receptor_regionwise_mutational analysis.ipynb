```python
import pandas as pd
import plotly.express as px
import plotly.graph_objects as go
```

```python
df = pd.read_csv('Receptors.csv')
df = df.rename(columns={'Receptor|Region': 'Receptor'})
```

```python
# 2. Reshape data into long format
df_melted = df.melt(id_vars=['Receptor'], var_name='Region', value_name='Mutations')
```

```python
fig = px.scatter(
    df_melted, 
    x='Region', 
    y='Mutations', 
    color='Receptor',
    title='Interactive Mutation Distribution across Receptor Regions',
    labels={'Region': 'Receptor Regions', 'Mutations': 'No. of Mutations'},
    hover_data={'Receptor': True, 'Region': True, 'Mutations': True}
)

# Update trace settings to make dots larger
fig.update_traces(marker=dict(size=10))

# 4. Enforce the Y-axis scale (1 step/unit = 2 mutations)
max_mutations = df_melted['Mutations'].max()
fig.update_layout(
    yaxis=dict(
        tickmode='linear',
        tick0=0,
        dtick=2,  # Forces grid lines and labels to tick exactly every 2 units
        range=[-0.5, max_mutations + 2]
    ),
    xaxis=dict(
        tickangle=45  # Keeps the structural region labels neat
    ),
    template='plotly_white',
    width=1100,
    height=650,
    legend=dict(
        title='<b>Receptors</b><br><dfn style="font-size:10px;color:gray;">Click a name to toggle</dfn>',
        orientation='v',
        yanchor='top',
        y=1,
        xanchor='left',
        x=1.02
    )
)

# 5. Save as an interactive HTML file
fig.write_html('interactive_receptor_plot.html')
print("Interactive plot saved successfully as 'interactive_receptor_plot.html'. Open this file in your browser!")
```
